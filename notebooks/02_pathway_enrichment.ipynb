```python
import pandas as pd
import gseapy as gp

print("Wczytywanie wyników DGE...")
df = pd.read_csv('../data/processed/deseq2_results_annotated.csv', index_col=0)

# Bierzemy geny ze zwiększoną ekspresją
up_genes = df[(df['padj'] < 0.05) & (df['log2FoldChange'] > 1)]['Gene_Symbol'].dropna().tolist()

print(f"Znaleziono {len(up_genes)} istotnie podwyższonych genów. Wysyłanie zapytania do baz Enrichr...")

# ZMIANA: Szukamy w KEGG oraz w GO Biological Process!
enr = gp.enrichr(
    gene_list=up_genes,
    gene_sets=['KEGG_2021_Human', 'GO_Biological_Process_2023'],
    organism='human',
    outdir=None 
)

enrichment_results = enr.results

# ZMIANA: Zamiast odrzucać, po prostu sortujemy po surowym P-value, aby zobaczyć najsilniejsze trendy
top_pathways = enrichment_results.sort_values('P-value').head(15)

# Wyświetlamy najważniejsze kolumny, w tym 'Gene_set' żeby wiedzieć z której bazy pochodzi wynik
display(top_pathways[['Gene_set', 'Term', 'Overlap', 'P-value', 'Adjusted P-value', 'Genes']])

# Zapisujemy plik dla dashboardu
top_pathways.to_csv('../data/processed/enrichment_upregulated.csv', index=False)
print("Zapisano top 15 szlaków biologicznych!")
```

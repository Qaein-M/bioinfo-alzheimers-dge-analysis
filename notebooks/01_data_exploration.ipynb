```python
import pandas as pd

# Ścieżka do Twojego pliku - upewnij się, że nazwa pliku się zgadza!
# Zwróć uwagę na parametr sep - pliki mogą być oddzielone przecinkiem (',') lub tabulatorem ('\t')
file_path = '/Users/Miko/Documents/Projekt_git/bioinfo-cancer-dge-analysis/data/raw/GSE203206_Subramaniam.ADRC_brain.counts.tsv' 

# Wczytujemy dane. index_col=0 sprawia, że nazwy genów będą indeksami wierszy
counts_df = pd.read_csv(file_path ,sep = '\t' , index_col=0)

# Wyświetlamy 5 pierwszych wierszy i wymiary tabeli
print(f"Wymiary macierzy (geny x próbki): {counts_df.shape}")
display(counts_df.head())
```

```python
# Sprawdzamy liczbę genów przed filtrowaniem
genes_before = counts_df.shape[0]

# Filtrujemy: zostawiamy tylko te wiersze (geny), 
# w których suma zliczeń ze wszystkich próbek jest większa niż 0
counts_df_filtered = counts_df.loc[(counts_df.sum(axis=1) > 0)]

# Sprawdzamy liczbę genów po filtrowaniu
genes_after = counts_df_filtered.shape[0]

print(f"Liczba genów przed filtrowaniem: {genes_before}")
print(f"Liczba genów po filtrowaniu: {genes_after}")
print(f"Usunięto {genes_before - genes_after} nieaktywnych genów.")

# Opcjonalnie: podgląd wyczyszczonych danych
display(counts_df_filtered.head())
```

```python
# Wyciągamy nazwy wszystkich próbek z nagłówków naszej wyczyszczonej macierzy zliczeń
sample_names = counts_df.columns.tolist()

# Tworzymy pustą listę na przypisanie do grup
conditions = []

# Pętla przypisująca próbki do grup na podstawie pierwszej litery (C, E, L)
for sample in sample_names:
    if sample.startswith('C_'):
        conditions.append('Control')
    else:
        # Grupujemy pacjentów 'E' i 'L' do jednego worka: 'Alzheimer'
        conditions.append('Alzheimer')

# Tworzymy nowy DataFrame z metadanymi dla naszego algorytmu
metadata_df = pd.DataFrame({
    'Sample': sample_names,
    'Condition': conditions
})

# BARDZO WAŻNE: W PyDESeq2 nazwy próbek w metadanych muszą być indeksem
metadata_df.set_index('Sample', inplace=True)

# Upewniamy się, że kolejność próbek w metadanych i macierzy zliczeń jest w 100% identyczna
# (bez tego algorytm statystyczny wyrzuci błąd)
counts_df = counts_df[metadata_df.index]

# Sprawdzamy efekt naszej pracy
print("Liczebność grup:")
print(metadata_df['Condition'].value_counts())
display(metadata_df.head(10))
```

```python
from pydeseq2.dds import DeseqDataSet
from pydeseq2.ds import DeseqStats

# Transpozycja macierzy
counts_df_transposed = counts_df.T

# 2. Inicjalizacja obiektu DeseqDataSet
print("Przygotowywanie danych do modelu...")
dds = DeseqDataSet(
    counts=counts_df_transposed,
    metadata=metadata_df,
    design="~Condition",  # ZMIANA: nowoczesny sposób definiowania podziału (formuła wzorowana na języku R)
    refit_cooks=True,
    n_cpus=1  # ZMIANA: wyłączamy wielowątkowość, co rozwiązuje błąd BrokenProcessPool
)

# 3. Dopasowanie modelu
print("Uruchamianie algorytmu DESeq2...")
dds.deseq2()

# 4. Obliczanie statystyk
print("Obliczanie statystyk dla kontrastu Alzheimer vs Control...")
stat_res = DeseqStats(
    dds, 
    contrast=["Condition", "Alzheimer", "Control"], 
    n_cpus=1  # ZMIANA: wielowątkowość na 1
)
stat_res.summary()

# 5. Wyciągnięcie wyników do formatu DataFrame
results_df = stat_res.results_df

# Podgląd pierwszych wyników
display(results_df.head())

# 6. Zapis wyników
results_df.to_csv('../data/processed/deseq2_results.csv')
```

```python
import plotly.express as px
import numpy as np

# 1. Usuwamy wiersze z wartościami pustymi (NaN), które algorytm odrzucił
df_plot = results_df.dropna(subset=['padj', 'log2FoldChange']).copy()

# 2. Obliczamy -log10 z p-value (standard na osi Y w Volcano Plot)
df_plot['-log10(padj)'] = -np.log10(df_plot['padj'])

# 3. Kategoryzujemy geny (Up-regulated, Down-regulated, Not significant)
# Przyjmujemy progi: skorygowane p-value < 0.05 oraz zmiana Fold Change min. 2-krotna (log2 > 1 lub < -1)
df_plot['Significance'] = 'Not significant'
df_plot.loc[(df_plot['padj'] < 0.05) & (df_plot['log2FoldChange'] > 1), 'Significance'] = 'Up-regulated'
df_plot.loc[(df_plot['padj'] < 0.05) & (df_plot['log2FoldChange'] < -1), 'Significance'] = 'Down-regulated'

# 4. Rysujemy interaktywny wykres
fig = px.scatter(
    df_plot, 
    x='log2FoldChange', 
    y='-log10(padj)',
    color='Significance',
    color_discrete_map={
        'Up-regulated': 'red', 
        'Down-regulated': 'blue', 
        'Not significant': 'lightgrey'
    },
    hover_name=df_plot.index,  # Nazwa genu po najechaniu myszką
    title='Volcano Plot: Alzheimer vs Control',
    labels={'log2FoldChange': 'Log2 Fold Change', '-log10(padj)': '-Log10 Adjusted P-value'}
)

# 5. Dodajemy linie progowe dla lepszej czytelności
fig.add_hline(y=-np.log10(0.05), line_dash="dash", line_color="black", annotation_text="padj = 0.05")
fig.add_vline(x=1, line_dash="dash", line_color="black")
fig.add_vline(x=-1, line_dash="dash", line_color="black")

# Wyświetlamy
fig.show()
```
